Import Modules and load dataframe

```python
from pathlib import Path
import sys

analysis_dir = Path.cwd().resolve()  # notebook lives in Analysis/
if str(analysis_dir) not in sys.path:
    sys.path.append(str(analysis_dir))

from results_loader import load_all_results
import analytics as ana
import plots as viz

df = load_all_results()     # auto-detects Scanner/Results
df = ana.prep_df(df)
```

Comliance over time by Location

```python
loc_trends = ana.compute_location_trends(df)
viz.plot_location_trends(loc_trends)
```

Compliance over time by Boundary

```python
boundary_trend = (
    df.groupby(["ScanDay", "Boundary"])["Status"]
      .value_counts()
      .unstack(fill_value=0)
      .sort_index()
)

# Pivot to make each boundary its own grouped bar
for boundary in df["Boundary"].unique():
    sub = (
        df[df["Boundary"] == boundary]
        .groupby("ScanDay")["Status"]
        .value_counts()
        .unstack(fill_value=0)
        .sort_index()
    )
    if sub.empty:
        continue
    viz.plot_stacked_counts(sub, f"Compliance Trend – Boundary: {boundary}", "Scan Date")
```
